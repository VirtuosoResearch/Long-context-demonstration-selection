```python
# %%
from cProfile import label
import matplotlib.pyplot as plt
import math
import numpy as np
import matplotlib as mpl
from matplotlib import rc
import seaborn as sns
rc('font',**{'family':'sans-serif','sans-serif':['Helvetica']})
mpl.rcParams['savefig.dpi'] = 1200
mpl.rcParams['text.usetex'] = True  # not really needed

# fill in data
drop = np.array([0.4, 0.55, 0.45, 0.3, 0.8, 0.2])
pp = np.mean(np.delete(drop, 0))
ti = np.mean(np.delete(drop, 1))
cs = np.mean(np.delete(drop, 2))
ops = np.mean(np.delete(drop, 3))
pmo = np.mean(np.delete(drop, 4))
ef = np.mean(np.delete(drop, 5))

print(pp, ti, cs, ops, pmo, ef)
colors = [
    '#3b80c2', '#5a98d0' , 
    '#7ab1da', '#829ad3', '#906cc4', '#9866bc',
    '#a95cae', '#ae57ab', '#b856a4', '#bd5aaf', '#c15db5', '#c763c0', '#cc68c6', '#d36cd1'
]



N = 1
ind = np.arange(N) * 24  # the x locations for the groups
width = 4.0      # the width of the bars
shift = 2.0


plt.rc('text', usetex=True)
plt.rc('font', family='serif')
fig, ax = plt.subplots(figsize=(8,5.5))
rects1 = ax.bar(ind, pp, width, color=colors[0], ecolor='white',) # color='yellowgreen', ecolor='k', hatch="|"
rects2 = ax.bar(ind + width * 1 + shift*1, ti, width, color=colors[1], ecolor='white') # color='tomato', ecolor='k', hatch="x"
rects3 = ax.bar(ind + width * 2 + shift*2, cs, width, color=colors[2], ecolor='white',)#, hatch='/'
rects4 = ax.bar(ind + width * 3 + shift*3, ops, width, color=colors[3], ecolor='white')
rects5 = ax.bar(ind + width * 4 + shift*4, pmo, width, color=colors[4], ecolor='white') # hatch="\\"
rects6 = ax.bar(ind + width * 5 + shift*5, ef, width, color=colors[5], ecolor='white') # hatch="\\"


#ax.set_ylim([0.2, 400])
ax.set_ylabel(r'$\mathrm{Attribution}$', fontsize=36)
ax.set_xticks([-0.5, 5, 12.5, 18, 24, 30])
ax.xaxis.set_ticks_position('none') 
ax.set_xticklabels([r'$\mathrm{PP}$', r'$\mathrm{TI}$', r'$\mathrm{CS}$', r'$\mathrm{OPS}$', r'$\mathrm{PMO}$', r'$\mathrm{EF}$'], rotation = 25)
plt.yticks([0.3, 0.4, 0.5], fontsize=36) #), [r"$0$", "", r"$2$", "", r"$4$", "", r"$6$"]
plt.ylim([0.25, 0.55])
plt.tick_params(axis='x')

ax.yaxis.grid(True, lw=0.8)
# ax.set_title('Top Singular Value of the Modified Weight Matrix', fontsize=32,  y=1.02)

ax.tick_params(axis='both', which='major', labelsize=36)
ax.tick_params(axis='both', which='minor', labelsize=36)

plt.title(r'$\mathrm{Gemma}-\mathrm{7B}$', fontsize=36)
plt.tight_layout()
plt.savefig('./gemma_attribution.pdf', format='pdf', dpi=1200)

```
